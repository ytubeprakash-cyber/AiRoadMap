Two app sessions each have three features. The weights turn each session into four scores. Predict the second session's scores before running the cell.


```python
# Import PyTorch for tensors and automatic gradients.
import torch

# Two app sessions, three features per session
# Create two rows of three numeric features.
X = torch.arange(1, 7, dtype=torch.float32).reshape(2, 3)

# Three input features, four output scores
# Start the matrix of feature weights for four outputs.
W = torch.tensor([
    # Map feature one to score one and the combined score.
    [1.0, 0.0, 0.0, 1.0],
    # Map feature two to score two and the combined score.
    [0.0, 1.0, 0.0, 1.0],
    # Map feature three to score three and the combined score.
    [0.0, 0.0, 1.0, 1.0],
# Match the weight matrix's type and device to the input.
], dtype=X.dtype, device=X.device)

# Give each output column its own constant offset.
bias = torch.tensor([10.0, 20.0, 30.0, 40.0], dtype=X.dtype, device=X.device)

# Multiply each input row by every weight column.
scores = X @ W
# Broadcast the same four offsets over both sessions.
output = scores + bias

# Display the input features.
print("Input:\n", X)
# Display raw weighted scores.
print("Before bias:\n", scores)
# Display the final scores.
print("After bias:\n", output)
# Show output dimensions and CPU or GPU placement.
print("Shape:", output.shape, "| Device:", output.device)

# Check that two sessions each have four output scores.
assert output.shape == (2, 4)
```

## How the shapes fit

`X @ W` multiplies a `(2, 3)` tensor by a `(3, 4)` tensor to give `(2, 4)`. The bias has shape `(4,)`. Broadcasting adds the same four bias values to each row.

For the first session, the weights give `[1, 2, 3, 6]`. Adding the bias gives `[11, 22, 33, 46]`.
