# Week 3 · Backpropagation by hand

A *computational graph* records how each result was calculated. Start with two numbers, multiply them, and square the result. Predict each value before running the next cell.

`a = 2` and `b = 3` → `c = a × b` → `loss = c²`.

```python
# Pick the two input values.
a = 2.0
b = 3.0

# Multiply the inputs to get the intermediate result.
c = a * b

# Square that intermediate result to get the final loss.
loss = c ** 2

# Show both values so we can check our arithmetic.
print(f"c = {c:g}")
print(f"loss = {loss:g}")
```

## Follow the arrows backward

Each local slope says how much one result changes when its immediate input changes a little.

1. `loss = c²`, so the slope of loss with respect to c is `2 × c = 12`.
2. `c = a × b`, so the slope of c with respect to a is `b = 3`, and with respect to b it is `a = 2`.
3. Multiply slopes along each route. The slope of loss with respect to a is `12 × 3 = 36`. The slope with respect to b is `12 × 2 = 24`.

This is the chain rule. The forward pass calculates values. The backward pass calculates how sensitive the final loss is to each earlier value.

```python
# The derivative of c squared with respect to c is 2 times c.
d_loss_d_c = 2 * c

# For c = a times b, changing a changes c at a rate of b.
d_c_d_a = b

# For c = a times b, changing b changes c at a rate of a.
d_c_d_b = a

# Multiply local slopes along the path from a to the loss.
d_loss_d_a = d_loss_d_c * d_c_d_a

# Multiply local slopes along the path from b to the loss.
d_loss_d_b = d_loss_d_c * d_c_d_b

# Display both input gradients.
print(f"Gradient for a: {d_loss_d_a:g}")
print(f"Gradient for b: {d_loss_d_b:g}")

# Confirm the expected gradients for this example.
assert d_loss_d_a == 36 and d_loss_d_b == 24
```
