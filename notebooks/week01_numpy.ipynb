# Week 1 — Vectors, matrices & shapes
Goal: prove my loop matmul matches NumPy, check shapes, and measure why vectorization matters.

```python
# Import NumPy for arrays, statistics, and matrix multiplication.
import numpy as np
# Load your own dot product and matrix multiplication functions.
from modelcraft_lab.linalg import dot, matmul

# Make random values reproducible when the cells run in the same order.
rng = np.random.default_rng(seed=0)
```

```python
# Create the first vector with three values.
a = [1.0, 2.0, 3.0]
# Create the second vector with three matching values.
b = [4.0, 5.0, 6.0]
# Compare your dot product with two NumPy calculations.
print(dot(a, b), np.dot(a, b), np.array(a) @ np.array(b))
```

```python
# Draw a matrix with two rows and three columns.
A = rng.normal(size=(2, 3))
# Draw a matrix with three rows and four columns.
B = rng.normal(size=(3, 4))
# Print the input shapes and the matrix product shape.
print("A:", A.shape, "B:", B.shape, "A @ B:", (A @ B).shape)

# Confirm that the outer dimensions form a 2 by 4 output.
assert (A @ B).shape == (2, 4)
```

```python
# Try an operation that may raise a shape error.
try:
    # Multiply incompatible shapes to demonstrate the error.
    A @ A
# Catch NumPy's shape error so the notebook continues.
except ValueError as e:
    # Show the error message about the mismatched inner dimensions.
    print("Shape error:", e)
```

```python
# Create a small 2 by 2 example with easy arithmetic.
X = np.array([[1, 2], [3, 4]])
# Square each element independently.
print("X * X =\n", X * X)
# Multiply each row by each column.
print("X @ X =\n", X @ X)
```

```python
# Create the first large matrix for timing.
big_A = rng.normal(size=(200, 200))
# Create the second large matrix for timing.
big_B = rng.normal(size=(200, 200))
# Convert NumPy arrays to nested Python lists for your implementation.
list_A, list_B = big_A.tolist(), big_B.tolist()

# Time repeated runs of the Python matrix multiplication.
%timeit matmul(list_A, list_B)
# Time repeated runs of NumPy matrix multiplication.
%timeit big_A @ big_B
```

## What I learned
Loop matmul: 200 ms. NumPy: 0.0385 ms. Speedup ≈ 5,200×
Why NumPy is faster: NumPy uses optimized compiled code for the calculations, which greatly reduces Python loop overhead.
The shape rule in one sentence: For (m, n) @ (p, q), the multiplication works when n == p, and the result has shape (m, q).

```python
# Import NumPy for arrays, statistics, and matrix multiplication.
import numpy as np

# Store five observed app startup times in milliseconds.
startup_ms = np.array([100, 150, 200, 250, 300])

# Report the average startup time.
print("Mean:", startup_ms.mean(), "ms")
# Report population variance, the mean squared distance from the average.
print("Variance:", startup_ms.var(), "ms²")
# Report spread in the original unit of milliseconds.
print("Standard deviation:", startup_ms.std(), "ms")
# Count True values as 1 and average them to get the observed fraction.
print("Fraction above 200 ms:", (startup_ms > 200).mean())
```

## Loss, gradient, and learning rate

The model predicts one value, `w`, and the target is 3. The squared error is `(w - 3) ** 2`. Its slope is `2 * (w - 3)`, which is −4 at `w = 1`. Subtracting the learning rate times this negative slope increases the weight.

Predict where the weight will land at learning rates of 0.1 and 1.1, then run the cell to compare the losses.

```python
# Define the loss function for a single model weight.
def loss(w: float) -> float:
    # Square the difference between the weight and target value 3.
    return (w - 3) ** 2


# Start the model weight at 1.
w = 1.0
# Calculate the slope of the loss at the current weight.
gradient = 2 * (w - 3)
# Update the weight using a learning rate of 0.1.
small_step = w - 0.1 * gradient
# Try a larger update that overshoots the target.
large_step = w - 1.1 * gradient

# Show the initial weight and loss.
print(f"Start: w = {w:.2f}, loss = {loss(w):.2f}")
# Show the result of the smaller update.
print(f"Rate 0.1: w = {small_step:.2f}, loss = {loss(small_step):.2f}")
# Show the result of the larger update.
print(f"Rate 1.1: w = {large_step:.2f}, loss = {loss(large_step):.2f}")

# Confirm that the small step helps and the large step hurts.
assert loss(small_step) < loss(w) < loss(large_step)
```

The 0.1 step moves the weight to 1.4 and lowers the loss from 4.00 to 2.56. The 1.1 step overshoots to 5.4 and raises the loss to 5.76.

In the startup example, the mean is 200 ms and the population variance is 5,000 ms². The standard deviation is about 70.71 ms, and two of five observed launches took strictly more than 200 ms.
