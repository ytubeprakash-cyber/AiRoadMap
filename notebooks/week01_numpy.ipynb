# Week 1 — Vectors, matrices & shapes
Goal: prove my loop matmul matches NumPy, check shapes, and measure why vectorization matters.

```python
import numpy as np
from modelcraft_lab.linalg import dot, matmul

rng = np.random.default_rng(seed=0)
```

```python
a = [1.0, 2.0, 3.0]
b = [4.0, 5.0, 6.0]
print(dot(a, b), np.dot(a, b), np.array(a) @ np.array(b))
```

```python
A = rng.normal(size=(2, 3))
B = rng.normal(size=(3, 4))
print("A:", A.shape, "B:", B.shape, "A @ B:", (A @ B).shape)

assert (A @ B).shape == (2, 4)
```

```python
try:
    A @ A
except ValueError as e:
    print("Shape error:", e)
```

```python
X = np.array([[1, 2], [3, 4]])
print("X * X =\n", X * X)
print("X @ X =\n", X @ X)
```

```python
big_A = rng.normal(size=(200, 200))
big_B = rng.normal(size=(200, 200))
list_A, list_B = big_A.tolist(), big_B.tolist()

%timeit matmul(list_A, list_B)
%timeit big_A @ big_B
```

## What I learned
Loop matmul: 200 ms. NumPy: 0.0385 ms. Speedup ≈ 5,200×
Why NumPy is faster: NumPy uses optimized compiled code for the calculations, which greatly reduces Python loop overhead.
The shape rule in one sentence: For (m, n) @ (p, q), the multiplication works when n == p, and the result has shape (m, q).

```python
import numpy as np

startup_ms = np.array([100, 150, 200, 250, 300])

print("Mean:", startup_ms.mean(), "ms")
print("Variance:", startup_ms.var(), "ms²")
print("Standard deviation:", startup_ms.std(), "ms")
print("Fraction above 200 ms:", (startup_ms > 200).mean())
```
